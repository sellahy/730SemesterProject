### Visual

```python
def plot_readings(df, imu: str, title: str, reading: str, dims: list[str] = ["x", "y", "z"]):
    fig, ax = plt.subplots()

    for dim in dims:
        ax.plot(
            df["timestamp"],
            df[f"{imu}_{reading}_{dim}"],
            label=dim
        )

    ax.set_title(title) # 'IMU 0 Linear Acceleration (Stationary)'
    ax.set_xlabel('timestamp')
    ax.set_ylabel(reading)
    #ax.xaxis.set_major_locator(mdates.SecondLocator(interval=1))
    fig.autofmt_xdate()
    fig.legend()

    plt.show()

plot_readings(
    circle_left_df,
    "Imu0",
    "IMU 0 Linear Acceleration (Circle Left)",
    "linear_accleration"
)
```

```python
plot_readings(
    circle_left_df,
    "Imu0",
    "IMU 0 Angular Velocity (Circle Left)",
    "angular_velocity"
)
```
